# Quebec Multi-Unit Property Analysis

This notebook provides an exploratory data analysis of multi-unit residential properties in Quebec, Canada. It includes financial calculations, scenario analysis, investment scoring, and report generation.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from quebec_multiunit.finance import calculate_cash_flow, calculate_roi
from quebec_multiunit.scenarios import analyze_scenarios
from quebec_multiunit.scoring import score_investment
from quebec_multiunit.reports import generate_report

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load property data
data_path = '../data/processed/sample_processed.csv'
properties_df = pd.read_csv(data_path)

# Display the first few rows of the dataset
properties_df.head()
```

```python
# Financial Analysis
properties_df['Cash Flow'] = properties_df.apply(lambda row: calculate_cash_flow(row['Income'], row['Expenses']), axis=1)
properties_df['ROI'] = properties_df.apply(lambda row: calculate_roi(row['Investment'], row['Cash Flow']), axis=1)

# Display financial metrics
financial_metrics = properties_df[['Property', 'Cash Flow', 'ROI']]
financial_metrics
```

```python
# Scenario Analysis
scenario_results = analyze_scenarios(properties_df)

# Display scenario analysis results
scenario_results
```

```python
# Investment Scoring
properties_df['Score'] = properties_df.apply(lambda row: score_investment(row['ROI'], row['Cash Flow']), axis=1)

# Display investment scores
investment_scores = properties_df[['Property', 'Score']]
investment_scores
```

```python
# Generate Report
report_path = '../reports/investment_analysis_report.pdf'
generate_report(properties_df, report_path)

# Report generated successfully
```
